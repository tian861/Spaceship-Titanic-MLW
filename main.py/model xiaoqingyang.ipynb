```python
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestClassifier
import os

# 加载数据
train = pd.read_csv("/kaggle/input/competitions/spaceship-titanic/train.csv")
test = pd.read_csv("/kaggle/input/competitions/spaceship-titanic/test.csv")
submission = pd.read_csv("/kaggle/input/competitions/spaceship-titanic/sample_submission.csv")

# 缺失值填充
train["Age"].fillna(train["Age"].median(), inplace=True)
test["Age"].fillna(train["Age"].median(), inplace=True)

train["CryoSleep"].fillna(False, inplace=True)
test["CryoSleep"].fillna(False, inplace=True)

train["VIP"].fillna(False, inplace=True)
test["VIP"].fillna(False, inplace=True)

cost_cols = ["RoomService", "FoodCourt", "ShoppingMall", "Spa", "VRDeck"]
for col in cost_cols:
    train[col].fillna(0, inplace=True)
    test[col].fillna(0, inplace=True)

# 布尔转数字
train["CryoSleep"] = train["CryoSleep"].astype(int)
test["CryoSleep"] = test["CryoSleep"].astype(int)
train["VIP"] = train["VIP"].astype(int)
test["VIP"] = test["VIP"].astype(int)

# 特征选择
features = ["Age", "CryoSleep", "VIP", "RoomService", "FoodCourt", "ShoppingMall", "Spa", "VRDeck"]
X = train[features]
y = train["Transported"]  # 现在这里就不会报错了
X_test = test[features]

# 训练模型
model = RandomForestClassifier(n_estimators=200, random_state=42)
model.fit(X, y)

# 预测
predictions = model.predict(X_test)

# 生成提交文件
submission["Transported"] = predictions
submission.to_csv("/kaggle/working/submission.csv", index=False)

print("✅ 文件生成成功！")
print("工作目录文件：", os.listdir("/kaggle/working/"))
```
